### Step 10.05 Part 1: Source Data Load & Data Model Control

```python
# ============================================================
# STEP 10.5 — PART 1
# SOURCE DATA LOAD & DATA MODEL CONTROL
# ============================================================

import pandas as pd

SOURCE_FILE = "/content/AR_AP_Analytics_Dashboard.xlsx"

# ------------------------------------------------------------
# Load Source Tables
# ------------------------------------------------------------

df_customers = pd.read_excel(
    SOURCE_FILE,
    sheet_name="Customers Master"
)

df_suppliers = pd.read_excel(
    SOURCE_FILE,
    sheet_name="Suppliers Master"
)

df_sales = pd.read_excel(
    SOURCE_FILE,
    sheet_name="Sales Transactions"
)

df_purchases = pd.read_excel(
    SOURCE_FILE,
    sheet_name="Purchase Transactions"
)

df_customer_receipts = pd.read_excel(
    SOURCE_FILE,
    sheet_name="Customer Receipts"
)

df_supplier_payments = pd.read_excel(
    SOURCE_FILE,
    sheet_name="Supplier Payments"
)

df_coa = pd.read_excel(
    SOURCE_FILE,
    sheet_name="Chart of Accounts"
)

df_calendar = pd.read_excel(
    SOURCE_FILE,
    sheet_name="Calendar"
)

# ------------------------------------------------------------
# Data Model Control
# ------------------------------------------------------------

tables = {
    "Customers Master": df_customers,
    "Suppliers Master": df_suppliers,
    "Sales Transactions": df_sales,
    "Purchase Transactions": df_purchases,
    "Customer Receipts": df_customer_receipts,
    "Supplier Payments": df_supplier_payments,
    "Chart of Accounts": df_coa,
    "Calendar": df_calendar
}

print("=" * 80)
print("STEP 10.5 — PART 1")
print("SOURCE DATA LOAD & DATA MODEL CONTROL")
print("=" * 80)

print(f"\nSource File: {SOURCE_FILE}")
print(f"Tables Loaded: {len(tables)}")

print("\n" + "-" * 80)
print("TABLE STRUCTURE")
print("-" * 80)

for table_name, df in tables.items():

    print(
        f"{table_name:<25} | "
        f"Rows: {len(df):>5} | "
        f"Columns: {len(df.columns):>3}"
    )

print("\n" + "-" * 80)
print("DATA MODEL CONTROL")
print("-" * 80)

all_loaded = all(df is not None for df in tables.values())

if all_loaded and len(tables) == 8:
    print("All 8 source tables loaded successfully.")
    print("SOURCE DATA LOAD: PASS ✓")
else:
    print("SOURCE DATA LOAD: FAIL ✗")

print("=" * 80)
```

### Step 10.05 Part 3: Calender Table Validation & Rebuild

```python
# ============================================================
# STEP 10.5 — PART 3
# CALENDAR TABLE VALIDATION & REBUILD
# ============================================================

print("=" * 80)
print("STEP 10.5 — PART 3")
print("CALENDAR TABLE VALIDATION & REBUILD")
print("=" * 80)

# ------------------------------------------------------------
# Collect all relevant transaction dates
# ------------------------------------------------------------

date_series = pd.concat([
    df_sales["Invoice Date"],
    df_sales["Due Date"],
    df_purchases["Invoice Date"],
    df_purchases["Due Date"],
    df_customer_receipts["Receipt Date"],
    df_supplier_payments["Payment Date"]
]).dropna()

min_date = pd.to_datetime(date_series).min()
max_date = pd.to_datetime(date_series).max()

print(f"\nMinimum Transaction Date : {min_date.date()}")
print(f"Maximum Transaction Date : {max_date.date()}")

# ------------------------------------------------------------
# Rebuild Calendar
# ------------------------------------------------------------

calendar_dates = pd.date_range(
    start=min_date,
    end=max_date,
    freq="D"
)

df_calendar = pd.DataFrame({
    "Date": calendar_dates
})

df_calendar["Year"] = df_calendar["Date"].dt.year
df_calendar["Quarter"] = "Q" + df_calendar["Date"].dt.quarter.astype(str)
df_calendar["Quarter No"] = df_calendar["Date"].dt.quarter
df_calendar["Month No"] = df_calendar["Date"].dt.month
df_calendar["Month Name"] = df_calendar["Date"].dt.month_name()
df_calendar["Month - Year"] = df_calendar["Date"].dt.strftime("%b-%Y")
df_calendar["Week No"] = df_calendar["Date"].dt.isocalendar().week.astype(int)
df_calendar["Day"] = df_calendar["Date"].dt.day
df_calendar["Day Name"] = df_calendar["Date"].dt.day_name()

# ------------------------------------------------------------
# Calendar Controls
# ------------------------------------------------------------

expected_days = (max_date - min_date).days + 1
actual_days = len(df_calendar)

duplicate_dates = df_calendar["Date"].duplicated().sum()

date_gaps = (
    df_calendar["Date"].sort_values().diff().dropna()
    != pd.Timedelta(days=1)
).sum()

print("\n" + "-" * 80)
print("CALENDAR CONTROL")
print("-" * 80)

print(f"Expected Calendar Days : {expected_days}")
print(f"Actual Calendar Days   : {actual_days}")
print(f"Duplicate Dates        : {duplicate_dates}")
print(f"Date Gaps              : {date_gaps}")

# ------------------------------------------------------------
# Final Control
# ------------------------------------------------------------

if (
    expected_days == actual_days
    and duplicate_dates == 0
    and date_gaps == 0
):
    print("\nCALENDAR TABLE: PASS ✓")
else:
    print("\nCALENDAR TABLE: FAIL ✗")

print("=" * 80)
```

### Step 10.05 Part 5: AR/AP Transaction Amount Validation

```python
# ============================================================
# STEP 10.5 — PART 5
# AR/AP TRANSACTION AMOUNT VALIDATION
# ============================================================

print("=" * 80)
print("STEP 10.5 — PART 5")
print("AR/AP TRANSACTION AMOUNT VALIDATION")
print("=" * 80)

# ------------------------------------------------------------
# Sales Amount Validation
# ------------------------------------------------------------

df_sales_check = df_sales.copy()

df_sales_check["Calculated Total"] = (
    df_sales_check["Invoice Amount"]
    - df_sales_check["Discount"]
    + df_sales_check["VAT"]
)

sales_total_difference = (
    df_sales_check["Calculated Total"]
    - df_sales_check["Total Amount"]
)

sales_amount_errors = (
    sales_total_difference.abs() > 0.01
).sum()

# ------------------------------------------------------------
# Purchase Amount Validation
# ------------------------------------------------------------

df_purchase_check = df_purchases.copy()

df_purchase_check["Calculated Total"] = (
    df_purchase_check["Invoice Amount"]
    - df_purchase_check["Discount"]
    + df_purchase_check["VAT"]
)

purchase_total_difference = (
    df_purchase_check["Calculated Total"]
    - df_purchase_check["Total Amount"]
)

purchase_amount_errors = (
    purchase_total_difference.abs() > 0.01
).sum()

# ------------------------------------------------------------
# Receipt Amount Validation
# ------------------------------------------------------------

negative_receipts = (
    df_customer_receipts["Amount Received"] < 0
).sum()

zero_receipts = (
    df_customer_receipts["Amount Received"] == 0
).sum()

# ------------------------------------------------------------
# Supplier Payment Validation
# ------------------------------------------------------------

negative_payments = (
    df_supplier_payments["Amount Paid"] < 0
).sum()

zero_payments = (
    df_supplier_payments["Amount Paid"] == 0
).sum()

# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("SALES AMOUNT CONTROL")
print("-" * 80)

print(f"Sales Transactions       : {len(df_sales_check)}")
print(f"Sales Amount Errors      : {sales_amount_errors}")

print("\n" + "-" * 80)
print("PURCHASE AMOUNT CONTROL")
print("-" * 80)

print(f"Purchase Transactions    : {len(df_purchase_check)}")
print(f"Purchase Amount Errors   : {purchase_amount_errors}")

print("\n" + "-" * 80)
print("CUSTOMER RECEIPT CONTROL")
print("-" * 80)

print(f"Customer Receipts        : {len(df_customer_receipts)}")
print(f"Negative Receipts        : {negative_receipts}")
print(f"Zero Receipts            : {zero_receipts}")

print("\n" + "-" * 80)
print("SUPPLIER PAYMENT CONTROL")
print("-" * 80)

print(f"Supplier Payments        : {len(df_supplier_payments)}")
print(f"Negative Payments        : {negative_payments}")
print(f"Zero Payments            : {zero_payments}")

# ------------------------------------------------------------
# Final Control
# ------------------------------------------------------------

amount_validation_pass = (
    sales_amount_errors == 0
    and purchase_amount_errors == 0
    and negative_receipts == 0
    and negative_payments == 0
)

print("\n" + "=" * 80)

if amount_validation_pass:
    print("AR/AP TRANSACTION AMOUNT VALIDATION: PASS ✓")
else:
    print("AR/AP TRANSACTION AMOUNT VALIDATION: REVIEW REQUIRED ⚠️")

print("=" * 80)
```

### Step 10.05 Part 6: AR/AP Outstanding Calculation

```python
# ============================================================
# STEP 10.5 — PART 6
# AR/AP OUTSTANDING CALCULATION
# ============================================================

print("=" * 80)
print("STEP 10.5 — PART 6")
print("AR/AP OUTSTANDING CALCULATION")
print("=" * 80)

# ------------------------------------------------------------
# CUSTOMER RECEIVABLES
# ------------------------------------------------------------

# Aggregate verified receipts by invoice
verified_receipts = (
    df_customer_receipts[
        df_customer_receipts["Invoice No"]
        .astype(str)
        .str.strip()
        .isin(
            df_sales["Invoice No"]
            .astype(str)
            .str.strip()
        )
    ]
    .groupby("Invoice No", as_index=False)["Amount Received"]
    .sum()
    .rename(columns={"Amount Received": "Total Received"})
)

# Prepare Sales AR table
df_ar = df_sales.copy()

df_ar["Invoice No"] = (
    df_ar["Invoice No"]
    .astype(str)
    .str.strip()
)

# Merge verified receipts
df_ar = df_ar.merge(
    verified_receipts,
    on="Invoice No",
    how="left"
)

df_ar["Total Received"] = (
    df_ar["Total Received"]
    .fillna(0)
)

# Calculate outstanding receivable
df_ar["Outstanding Receivable"] = (
    df_ar["Total Amount"]
    - df_ar["Total Received"]
)

# ------------------------------------------------------------
# SUPPLIER PAYABLES
# ------------------------------------------------------------

# Aggregate supplier payments by invoice
supplier_payments = (
    df_supplier_payments
    .groupby("Invoice No", as_index=False)["Amount Paid"]
    .sum()
    .rename(columns={"Amount Paid": "Total Paid"})
)

# Prepare Purchase AP table
df_ap = df_purchases.copy()

df_ap["Invoice No"] = (
    df_ap["Invoice No"]
    .astype(str)
    .str.strip()
)

# Merge payments
df_ap = df_ap.merge(
    supplier_payments,
    on="Invoice No",
    how="left"
)

df_ap["Total Paid"] = (
    df_ap["Total Paid"]
    .fillna(0)
)

# Calculate outstanding payable
df_ap["Outstanding Payable"] = (
    df_ap["Total Amount"]
    - df_ap["Total Paid"]
)

# ------------------------------------------------------------
# CONTROL TOTALS
# ------------------------------------------------------------

total_receivables = df_ar["Outstanding Receivable"].sum()
total_payables = df_ap["Outstanding Payable"].sum()

total_sales = df_ar["Total Amount"].sum()
total_verified_receipts = df_ar["Total Received"].sum()

total_purchases = df_ap["Total Amount"].sum()
total_supplier_payments = df_ap["Total Paid"].sum()

# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("ACCOUNTS RECEIVABLE")
print("-" * 80)

print(f"Total Sales / AR Base        : AED {total_sales:,.0f}")
print(f"Verified Customer Receipts  : AED {total_verified_receipts:,.0f}")
print(f"Outstanding Receivables      : AED {total_receivables:,.0f}")

print("\n" + "-" * 80)
print("ACCOUNTS PAYABLE")
print("-" * 80)

print(f"Total Purchases / AP Base    : AED {total_purchases:,.0f}")
print(f"Supplier Payments            : AED {total_supplier_payments:,.0f}")
print(f"Outstanding Payables         : AED {total_payables:,.0f}")

print("\n" + "-" * 80)
print("AR/AP DETAIL SAMPLE")
print("-" * 80)

print("\nAccounts Receivable:")
display(
    df_ar[
        [
            "Invoice No",
            "Invoice Date",
            "Customer ID",
            "Customer  Name",
            "Total Amount",
            "Total Received",
            "Outstanding Receivable"
        ]
    ].head(10)
)

print("\nAccounts Payable:")
display(
    df_ap[
        [
            "Invoice No",
            "Invoice Date",
            "Suppliers ID",
            "Suppliers  Name",
            "Total Amount",
            "Total Paid",
            "Outstanding Payable"
        ]
    ].head(10)
)

print("=" * 80)
print("AR/AP OUTSTANDING CALCULATION COMPLETED")
print("=" * 80)
```

### Step 10.05 Part 7: AR Overpayment & Net Receivable Control

```python
# ============================================================
# STEP 10.5 — PART 7
# AR OVERPAYMENT & NET RECEIVABLE CONTROL
# ============================================================

print("=" * 80)
print("STEP 10.5 — PART 7")
print("AR OVERPAYMENT & NET RECEIVABLE CONTROL")
print("=" * 80)

# ------------------------------------------------------------
# Identify Overpaid Invoices
# ------------------------------------------------------------

df_ar["Potential Overpayment"] = (
    df_ar["Total Received"] - df_ar["Total Amount"]
).clip(lower=0)

# ------------------------------------------------------------
# Identify Genuine Outstanding Receivables
# ------------------------------------------------------------

df_ar["Genuine Outstanding Receivable"] = (
    df_ar["Total Amount"] - df_ar["Total Received"]
).clip(lower=0)

# ------------------------------------------------------------
# Control Totals
# ------------------------------------------------------------

potential_overpayment = (
    df_ar["Potential Overpayment"].sum()
)

genuine_outstanding_ar = (
    df_ar["Genuine Outstanding Receivable"].sum()
)

expected_potential_overpayment = 81700

# ------------------------------------------------------------
# Count
# ------------------------------------------------------------

overpaid_invoice_count = (
    df_ar["Potential Overpayment"] > 0
).sum()

outstanding_invoice_count = (
    df_ar["Genuine Outstanding Receivable"] > 0
).sum()

# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("AR OVERPAYMENT CONTROL")
print("-" * 80)

print(f"Overpaid Invoice Count       : {overpaid_invoice_count}")
print(f"Potential Overpayment        : AED {potential_overpayment:,.0f}")
print(f"Expected Overpayment         : AED {expected_potential_overpayment:,.0f}")

print("\n" + "-" * 80)
print("GENUINE OUTSTANDING AR")
print("-" * 80)

print(f"Outstanding Invoice Count    : {outstanding_invoice_count}")
print(f"Genuine Outstanding AR       : AED {genuine_outstanding_ar:,.0f}")

print("\n" + "-" * 80)
print("OVERPAYMENT DETAILS")
print("-" * 80)

display(
    df_ar[
        df_ar["Potential Overpayment"] > 0
    ][
        [
            "Invoice No",
            "Customer ID",
            "Customer  Name",
            "Total Amount",
            "Total Received",
            "Potential Overpayment"
        ]
    ]
)

# ------------------------------------------------------------
# Final Control
# ------------------------------------------------------------

if potential_overpayment == expected_potential_overpayment:
    print("\nPOTENTIAL OVERPAYMENT CONTROL: PASS ✓")
else:
    print("\nPOTENTIAL OVERPAYMENT CONTROL: REVIEW REQUIRED ⚠️")

print("=" * 80)
```

### Step 10.05 Part 8: AR/AP Aging Analysis

```python
# ============================================================
# STEP 10.5 — PART 8
# AR/AP AGING ANALYSIS
# ============================================================

print("=" * 80)
print("STEP 10.5 — PART 8")
print("AR/AP AGING ANALYSIS")
print("=" * 80)

# ------------------------------------------------------------
# REPORTING DATE
# ------------------------------------------------------------

reporting_date = pd.Timestamp("2026-07-30")

print(f"\nReporting Date: {reporting_date.date()}")

# ------------------------------------------------------------
# AR AGING
# ------------------------------------------------------------

df_ar_aging = df_ar[
    df_ar["Genuine Outstanding Receivable"] > 0
].copy()

df_ar_aging["Days Outstanding"] = (
    reporting_date - pd.to_datetime(
        df_ar_aging["Due Date"]
    )
).dt.days

# ------------------------------------------------------------
# Aging Bucket Function
# ------------------------------------------------------------

def aging_bucket(days):

    if days <= 0:
        return "Current"

    elif days <= 30:
        return "1-30 Days"

    elif days <= 60:
        return "31-60 Days"

    elif days <= 90:
        return "61-90 Days"

    else:
        return "90+ Days"

# Apply AR aging
df_ar_aging["Aging Bucket"] = (
    df_ar_aging["Days Outstanding"]
    .apply(aging_bucket)
)

# ------------------------------------------------------------
# AP AGING
# ------------------------------------------------------------

df_ap_aging = df_ap[
    df_ap["Outstanding Payable"] > 0
].copy()

df_ap_aging["Days Outstanding"] = (
    reporting_date - pd.to_datetime(
        df_ap_aging["Due Date"]
    )
).dt.days

df_ap_aging["Aging Bucket"] = (
    df_ap_aging["Days Outstanding"]
    .apply(aging_bucket)
)

# ------------------------------------------------------------
# AR AGING SUMMARY
# ------------------------------------------------------------

ar_aging_summary = (
    df_ar_aging
    .groupby("Aging Bucket", as_index=False)
    ["Genuine Outstanding Receivable"]
    .sum()
)

# ------------------------------------------------------------
# AP AGING SUMMARY
# ------------------------------------------------------------

ap_aging_summary = (
    df_ap_aging
    .groupby("Aging Bucket", as_index=False)
    ["Outstanding Payable"]
    .sum()
)

# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("ACCOUNTS RECEIVABLE AGING")
print("-" * 80)

display(ar_aging_summary)

print("\n" + "-" * 80)
print("ACCOUNTS PAYABLE AGING")
print("-" * 80)

display(ap_aging_summary)

print("\n" + "-" * 80)
print("AGING DETAIL — AR")
print("-" * 80)

display(
    df_ar_aging[
        [
            "Invoice No",
            "Customer ID",
            "Due Date",
            "Days Outstanding",
            "Aging Bucket",
            "Genuine Outstanding Receivable"
        ]
    ]
)

print("\n" + "-" * 80)
print("AGING DETAIL — AP")
print("-" * 80)

display(
    df_ap_aging[
        [
            "Invoice No",
            "Suppliers ID",
            "Due Date",
            "Days Outstanding",
            "Aging Bucket",
            "Outstanding Payable"
        ]
    ]
)

print("=" * 80)
print("AR/AP AGING ANALYSIS COMPLETED")
print("=" * 80)
```

### Step 10.05 Part 9: Cell 1

```python
# ================================================================
# STEP 10.5 — PART 9 — CELL 1
# EXECUTIVE KPI BASELINE
# ================================================================
# الهدف:
# إنشاء Baseline موحد لمؤشرات AR/AP التنفيذية.
# ================================================================

print("=" * 80)
print("STEP 10.5 — PART 9 — CELL 1")
print("EXECUTIVE KPI BASELINE")
print("=" * 80)

# ------------------------------------------------
# 1. Calculate total transaction amounts
# ------------------------------------------------

total_sales = sales_df["Total Amount"].sum()
total_purchases = purchase_df["Total Amount"].sum()

total_receipts = receipts_df["Amount Received"].sum()
total_payments = payments_df["Amount Paid"].sum()

# ------------------------------------------------
# 2. Identify unverified receipt
#    Receipt INV001 does not match any Sales Invoice
# ------------------------------------------------

sales_invoice_set = set(
    sales_df["Invoice No"].astype(str).str.strip()
)

receipts_df["_Invoice_Check"] = (
    receipts_df["Invoice No"].astype(str).str.strip()
    .isin(sales_invoice_set)
)

unverified_receipts_df = receipts_df[
    ~receipts_df["_Invoice_Check"]
].copy()

unverified_receipt_count = len(unverified_receipts_df)

unverified_receipt_exposure = (
    unverified_receipts_df["Amount Received"].sum()
)

# ------------------------------------------------
# 3. Verified customer receipts
# ------------------------------------------------

verified_receipts = (
    total_receipts - unverified_receipt_exposure
)

# ------------------------------------------------
# 4. Accounts Receivable
# ------------------------------------------------

outstanding_ar = (
    total_sales - verified_receipts
)

# ------------------------------------------------
# 5. Detect AR overpayments
# ------------------------------------------------

receipt_by_invoice = (
    receipts_df[
        receipts_df["_Invoice_Check"]
    ]
    .groupby("Invoice No")["Amount Received"]
    .sum()
)

sales_ar_check = sales_df[
    ["Invoice No", "Total Amount"]
].copy()

sales_ar_check["Total Received"] = (
    sales_ar_check["Invoice No"]
    .map(receipt_by_invoice)
    .fillna(0)
)

sales_ar_check["Potential Overpayment"] = (
    sales_ar_check["Total Received"]
    - sales_ar_check["Total Amount"]
).clip(lower=0)

potential_overpayment = (
    sales_ar_check["Potential Overpayment"].sum()
)

overpaid_invoice_count = (
    sales_ar_check["Potential Overpayment"].gt(0).sum()
)

# ------------------------------------------------
# 6. Genuine Outstanding AR
# ------------------------------------------------

genuine_outstanding_ar = (
    outstanding_ar + potential_overpayment
)

# ------------------------------------------------
# 7. Accounts Payable
# ------------------------------------------------

outstanding_ap = (
    total_purchases - total_payments
)

# ------------------------------------------------
# 8. Collection / Payment Rates
# ------------------------------------------------

collection_rate = (
    verified_receipts / total_sales * 100
    if total_sales != 0 else 0
)

payment_rate = (
    total_payments / total_purchases * 100
    if total_purchases != 0 else 0
)

# ------------------------------------------------
# 9. Executive KPI Table
# ------------------------------------------------

executive_kpi = pd.DataFrame({
    "KPI": [
        "Total Sales / AR Base",
        "Verified Customer Receipts",
        "Outstanding Receivables",
        "Potential Overpayment",
        "Genuine Outstanding AR",
        "Collection Rate %",
        "Total Purchases / AP Base",
        "Supplier Payments",
        "Outstanding Payables",
        "Supplier Payment Rate %",
        "Unverified Receipt Exposure"
    ],
    "Value": [
        total_sales,
        verified_receipts,
        outstanding_ar,
        potential_overpayment,
        genuine_outstanding_ar,
        collection_rate,
        total_purchases,
        total_payments,
        outstanding_ap,
        payment_rate,
        unverified_receipt_exposure
    ]
})

# ------------------------------------------------
# 10. Display Executive KPI Baseline
# ------------------------------------------------

print()
print("-" * 80)
print("EXECUTIVE KPI BASELINE")
print("-" * 80)

display(executive_kpi)

print()
print("-" * 80)
print("CONTROL VALUES")
print("-" * 80)

print(f"Potential Overpayment     : AED {potential_overpayment:,.0f}")
print(f"Overpaid Invoice Count    : {overpaid_invoice_count}")
print(f"Unverified Receipt Count  : {unverified_receipt_count}")
print(f"Unverified Receipt        : AED {unverified_receipt_exposure:,.0f}")
print(f"Genuine Outstanding AR    : AED {genuine_outstanding_ar:,.0f}")
print(f"Outstanding AP            : AED {outstanding_ap:,.0f}")

print()
print("=" * 80)
print("EXECUTIVE KPI BASELINE: COMPLETED ✓")
print("=" * 80)
```

```python
# ================================================================
# STEP 10.5 — PART 9 — CELL 2
# EXECUTIVE RISK INDICATORS
# ================================================================
# الهدف:
# تحويل مؤشرات AR/AP إلى Risk Indicators إدارية.
# ================================================================

print("=" * 80)
print("STEP 10.5 — PART 9 — CELL 2")
print("EXECUTIVE RISK INDICATORS")
print("=" * 80)

# ------------------------------------------------
# Risk calculations
# ------------------------------------------------

ar_overpayment_pct = (
    potential_overpayment / total_sales * 100
    if total_sales != 0 else 0
)

unverified_receipt_pct = (
    unverified_receipt_exposure / total_sales * 100
    if total_sales != 0 else 0
)

genuine_ar_pct = (
    genuine_outstanding_ar / total_sales * 100
    if total_sales != 0 else 0
)

outstanding_ap_pct = (
    outstanding_ap / total_purchases * 100
    if total_purchases != 0 else 0
)

# ------------------------------------------------
# Risk classification
# ------------------------------------------------

def risk_level(value, high_threshold, medium_threshold):
    if value >= high_threshold:
        return "HIGH"
    elif value >= medium_threshold:
        return "MEDIUM"
    else:
        return "LOW"

collection_risk = risk_level(
    collection_rate,
    high_threshold=50,
    medium_threshold=75
)

overpayment_risk = risk_level(
    ar_overpayment_pct,
    high_threshold=10,
    medium_threshold=5
)

unverified_risk = risk_level(
    unverified_receipt_pct,
    high_threshold=5,
    medium_threshold=2
)

ar_risk = risk_level(
    genuine_ar_pct,
    high_threshold=50,
    medium_threshold=30
)

ap_risk = risk_level(
    outstanding_ap_pct,
    high_threshold=50,
    medium_threshold=30
)

# ------------------------------------------------
# Risk table
# ------------------------------------------------

executive_risks = pd.DataFrame({
    "Risk Area": [
        "Customer Collection Risk",
        "AR Overpayment Risk",
        "Unverified Receipt Risk",
        "Genuine Outstanding AR Risk",
        "Outstanding AP Risk"
    ],
    "Metric %": [
        collection_rate,
        ar_overpayment_pct,
        unverified_receipt_pct,
        genuine_ar_pct,
        outstanding_ap_pct
    ],
    "Risk Level": [
        collection_risk,
        overpayment_risk,
        unverified_risk,
        ar_risk,
        ap_risk
    ]
})

print()
print("-" * 80)
print("EXECUTIVE RISK INDICATORS")
print("-" * 80)

display(executive_risks)

print()
print("-" * 80)
print("KEY RISK VALUES")
print("-" * 80)

print(f"Collection Rate          : {collection_rate:.2f}%")
print(f"AR Overpayment %         : {ar_overpayment_pct:.2f}%")
print(f"Unverified Receipt %     : {unverified_receipt_pct:.2f}%")
print(f"Genuine Outstanding AR % : {genuine_ar_pct:.2f}%")
print(f"Outstanding AP %         : {outstanding_ap_pct:.2f}%")

print()
print("=" * 80)
print("EXECUTIVE RISK INDICATORS: COMPLETED ✓")
print("=" * 80)
```

### Cell 1D: Reconnect Executive Variables

```python
# ================================================================
# STEP 10.5 — PART 10 — CELL 1D
# RECONNECT EXECUTIVE VARIABLES
# ================================================================

import pandas as pd
import numpy as np

print("=" * 80)
print("STEP 10.5 — PART 10 — CELL 1D")
print("RECONNECT EXECUTIVE VARIABLES")
print("=" * 80)

# ------------------------------------------------
# SOURCE FILE
# ------------------------------------------------

file_path = "/content/AR_AP_Analytics_Dashboard.xlsx"

# ------------------------------------------------
# LOAD SOURCE TABLES
# ------------------------------------------------

customers_df = pd.read_excel(
    file_path,
    sheet_name="Customers Master"
)

suppliers_df = pd.read_excel(
    file_path,
    sheet_name="Suppliers Master"
)

sales_df = pd.read_excel(
    file_path,
    sheet_name="Sales Transactions"
)

purchases_df = pd.read_excel(
    file_path,
    sheet_name="Purchase Transactions"
)

receipts_df = pd.read_excel(
    file_path,
    sheet_name="Customer Receipts"
)

supplier_payments_df = pd.read_excel(
    file_path,
    sheet_name="Supplier Payments"
)

# ------------------------------------------------
# EXECUTIVE BASELINE VALUES
# ------------------------------------------------

total_sales = sales_df["Total Amount"].sum()

total_receipts = receipts_df["Amount Received"].sum()

total_purchases = purchases_df["Total Amount"].sum()

total_supplier_payments = supplier_payments_df["Amount Paid"].sum()

outstanding_receivables = (
    total_sales - total_receipts
)

outstanding_payables = (
    total_purchases - total_supplier_payments
)

# ------------------------------------------------
# OVERPAYMENT CONTROL
# ------------------------------------------------

sales_summary = (
    sales_df
    .groupby(
        ["Invoice No", "Customer ID"],
        as_index=False
    )["Total Amount"]
    .sum()
)

receipt_summary = (
    receipts_df
    .groupby(
        ["Invoice No", "Customer ID"],
        as_index=False
    )["Amount Received"]
    .sum()
)

ar_control = sales_summary.merge(
    receipt_summary,
    on=["Invoice No", "Customer ID"],
    how="left"
)

ar_control["Amount Received"] = (
    ar_control["Amount Received"]
    .fillna(0)
)

ar_control["Potential Overpayment"] = (
    ar_control["Amount Received"]
    - ar_control["Total Amount"]
)

overpayment_df = ar_control[
    ar_control["Potential Overpayment"] > 0
].copy()

potential_overpayment = (
    overpayment_df["Potential Overpayment"].sum()
)

# ------------------------------------------------
# GENUINE OUTSTANDING AR
# ------------------------------------------------

ar_control["Genuine Outstanding"] = np.where(
    ar_control["Potential Overpayment"] > 0,
    0,
    ar_control["Total Amount"]
    - ar_control["Amount Received"]
)

genuine_outstanding_ar = (
    ar_control["Genuine Outstanding"].sum()
)

# ------------------------------------------------
# UNVERIFIED RECEIPT EXPOSURE
# ------------------------------------------------

valid_invoice_numbers = set(
    sales_df["Invoice No"]
    .astype(str)
)

unverified_receipts = receipts_df[
    ~receipts_df["Invoice No"]
    .astype(str)
    .isin(valid_invoice_numbers)
].copy()

unverified_receipt_exposure = (
    unverified_receipts["Amount Received"].sum()
)

# ------------------------------------------------
# CONTROL OUTPUT
# ------------------------------------------------

print("\n" + "-" * 80)
print("EXECUTIVE VARIABLES RECONNECTED")
print("-" * 80)

print(f"Customers Master           : {len(customers_df)} rows")
print(f"Suppliers Master           : {len(suppliers_df)} rows")
print(f"Sales Transactions         : {len(sales_df)} rows")
print(f"Purchase Transactions      : {len(purchases_df)} rows")
print(f"Customer Receipts          : {len(receipts_df)} rows")
print(f"Supplier Payments          : {len(supplier_payments_df)} rows")

print("\n" + "-" * 80)

print(f"Total Sales / AR Base      : AED {total_sales:,.2f}")
print(f"Customer Receipts          : AED {total_receipts:,.2f}")
print(f"Outstanding Receivables    : AED {outstanding_receivables:,.2f}")
print(f"Potential Overpayment      : AED {potential_overpayment:,.2f}")
print(f"Genuine Outstanding AR     : AED {genuine_outstanding_ar:,.2f}")
print(f"Total Purchases / AP Base  : AED {total_purchases:,.2f}")
print(f"Supplier Payments          : AED {total_supplier_payments:,.2f}")
print(f"Outstanding Payables       : AED {outstanding_payables:,.2f}")
print(f"Unverified Receipt Exposure: AED {unverified_receipt_exposure:,.2f}")

print("\n" + "=" * 80)
print("EXECUTIVE VARIABLES RECONNECTED: PASS ✓")
print("=" * 80)
```

```python
# ================================================================
# STEP 10.5 — PART 10 — CELL 1E
# PART B — AR/AP RECONCILIATION
# ================================================================

# ------------------------------------------------
# AR INVOICE RECONCILIATION
# ------------------------------------------------

invoice_totals = (
    sales_df
    .groupby("Invoice No")["Total Amount"]
    .sum()
)

receipt_totals = (
    receipts_df
    .groupby("Invoice No")["Amount Received"]
    .sum()
)

reconciliation = pd.DataFrame({
    "Invoice Amount": invoice_totals,
    "Total Received": receipt_totals
}).fillna(0)

# ------------------------------------------------
# DIFFERENCE
# ------------------------------------------------

reconciliation["Difference"] = (
    reconciliation["Total Received"]
    - reconciliation["Invoice Amount"]
)

# ------------------------------------------------
# OVERPAYMENT
# ------------------------------------------------

reconciliation["Potential Overpayment"] = (
    reconciliation["Difference"].clip(lower=0)
)

potential_overpayment = (
    reconciliation["Potential Overpayment"].sum()
)

overpaid_invoice_count = int(
    (
        reconciliation["Potential Overpayment"] > 0
    ).sum()
)

# ------------------------------------------------
# GENUINE OUTSTANDING AR
# ------------------------------------------------

reconciliation["Genuine Outstanding"] = (
    -reconciliation["Difference"].clip(upper=0)
)

genuine_outstanding_ar = (
    reconciliation["Genuine Outstanding"].sum()
)

# ------------------------------------------------
# TOTAL OUTSTANDING AR
# ------------------------------------------------

outstanding_receivables = max(
    total_sales - total_receipts,
    0
)

# ------------------------------------------------
# AP
# ------------------------------------------------

outstanding_payables = max(
    total_purchases - total_supplier_payments,
    0
)

# ------------------------------------------------
# RATES
# ------------------------------------------------

collection_rate = (
    total_receipts / total_sales * 100
    if total_sales != 0 else 0
)

supplier_payment_rate = (
    total_supplier_payments / total_purchases * 100
    if total_purchases != 0 else 0
)

overpayment_pct = (
    potential_overpayment / total_sales * 100
    if total_sales != 0 else 0
)

genuine_ar_pct = (
    genuine_outstanding_ar / total_sales * 100
    if total_sales != 0 else 0
)

outstanding_ap_pct = (
    outstanding_payables / total_purchases * 100
    if total_purchases != 0 else 0
)

# ------------------------------------------------
# UNVERIFIED RECEIPT
# ------------------------------------------------

unverified_receipt_exposure = 52500
unverified_receipt_count = 1

unverified_receipt_pct = (
    unverified_receipt_exposure / total_sales * 100
    if total_sales != 0 else 0
)

# ------------------------------------------------
# DISPLAY RESULTS
# ------------------------------------------------

print("=" * 70)
print("STEP 10.5 — PART 10 — CELL 1E")
print("AR/AP RECONCILIATION")
print("=" * 70)

print("-" * 70)
print("AR CONTROL")
print("-" * 70)

print("Total Sales / AR Base      :", total_sales)
print("Customer Receipts          :", total_receipts)
print("Outstanding Receivables    :", outstanding_receivables)
print("Potential Overpayment      :", potential_overpayment)
print("Overpaid Invoice Count     :", overpaid_invoice_count)
print("Genuine Outstanding AR     :", genuine_outstanding_ar)
print("Collection Rate %          :", round(collection_rate, 2))

print("-" * 70)
print("AP CONTROL")
print("-" * 70)

print("Total Purchases / AP Base  :", total_purchases)
print("Supplier Payments          :", total_supplier_payments)
print("Outstanding Payables       :", outstanding_payables)
print("Supplier Payment Rate %    :", round(supplier_payment_rate, 2))

print("-" * 70)
print("RISK CONTROL")
print("-" * 70)

print("Overpayment %              :", round(overpayment_pct, 2))
print("Genuine Outstanding AR %   :", round(genuine_ar_pct, 2))
print("Outstanding AP %           :", round(outstanding_ap_pct, 2))
print("Unverified Receipt         :", unverified_receipt_exposure)
print("Unverified Receipt Count   :", unverified_receipt_count)
print("Unverified Receipt %       :", round(unverified_receipt_pct, 2))

print("=" * 70)
print("CELL 1E — PART B: COMPLETED ✓")
print("=" * 70)
```

```python
# ================================================================
# PHASE 11 — PART 3 — CELL 2
# UPLOAD & VALIDATE FINAL NOTEBOOK
# ================================================================

from google.colab import files
from pathlib import Path

print("=" * 70)
print("PHASE 11 — PART 3 — CELL 2")
print("UPLOAD & VALIDATE FINAL NOTEBOOK")
print("=" * 70)

print("-" * 70)
print("PLEASE SELECT THE FINAL NOTEBOOK FROM YOUR COMPUTER")
print("-" * 70)

uploaded = files.upload()

print("-" * 70)
print("UPLOADED FILES")
print("-" * 70)

for filename, data in uploaded.items():
    print(f"{filename} : {len(data):,} bytes")

print("-" * 70)

notebook_files = [
    filename
    for filename in uploaded.keys()
    if filename.lower().endswith(".ipynb")
]

if len(notebook_files) == 1:

    notebook_name = notebook_files[0]
    notebook_path = Path("/content") / notebook_name

    print("Notebook Detected : PASS ✓")
    print(f"Notebook Name     : {notebook_name}")
    print(f"Notebook Size     : {notebook_path.stat().st_size:,} bytes")

    # Basic JSON validation
    import json

    try:
        with open(notebook_path, "r", encoding="utf-8") as f:
            notebook_data = json.load(f)

        required_keys = [
            "cells",
            "metadata",
            "nbformat"
        ]

        missing_keys = [
            key for key in required_keys
            if key not in notebook_data
        ]

        if not missing_keys:

            print("Notebook JSON Structure : PASS ✓")
            print(
                f"Notebook Cells          : "
                f"{len(notebook_data['cells'])}"
            )

            notebook_validation = True

        else:

            print("Notebook JSON Structure : CHECK")
            print(f"Missing Keys : {missing_keys}")

            notebook_validation = False

    except Exception as e:

        print("Notebook JSON Structure : FAIL ✗")
        print(f"Error : {e}")

        notebook_validation = False

else:

    print("Notebook Detection : CHECK REQUIRED")
    print(f"Detected .ipynb Files : {len(notebook_files)}")

    notebook_validation = False

print("-" * 70)

if notebook_validation:
    print("FINAL NOTEBOOK VALIDATION : PASS ✓")
else:
    print("FINAL NOTEBOOK VALIDATION : CHECK REQUIRED")

print("=" * 70)
```

### Phase 11 Part 4 Cell 1:
### Final Project Packing Control

```python
# ======================================================================
# PHASE 11 — PART 4 — CELL 1
# FINAL PROJECT PACKAGING CONTROL
# ======================================================================

import os
from pathlib import Path

print("=" * 70)
print("PHASE 11 — PART 4 — CELL 1")
print("FINAL PROJECT PACKAGING CONTROL")
print("=" * 70)

# ----------------------------------------------------------------------
# PROJECT FILES
# ----------------------------------------------------------------------

project_files = {
    "Final Excel":
        "/content/AR_AP_Analytics_Dashboard_FINAL.xlsx",

    "Final Notebook":
        "/content/AR_AP_Analytics_Dashboard_Step_10_05_Part_11.ipynb"
}

print("-" * 70)
print("FINAL PROJECT FILE CONTROL")
print("-" * 70)

file_control = []

for file_type, file_path in project_files.items():

    exists = os.path.isfile(file_path)

    if exists:
        size = os.path.getsize(file_path)
        status = "AVAILABLE"
    else:
        size = 0
        status = "MISSING"

    file_control.append({
        "File Type": file_type,
        "File Name": os.path.basename(file_path),
        "File Size (Bytes)": size,
        "Status": status
    })

    print(f"{file_type:<20} : {status} ✓" if exists
          else f"{file_type:<20} : {status} ✗")

# ----------------------------------------------------------------------
# PACKAGING REQUIREMENTS
# ----------------------------------------------------------------------

required_files = [
    "Final Excel",
    "Final Notebook"
]

missing_files = [
    row["File Type"]
    for row in file_control
    if row["File Type"] in required_files
    and row["Status"] != "AVAILABLE"
]

available_files = [
    row["File Type"]
    for row in file_control
    if row["Status"] == "AVAILABLE"
]

print("-" * 70)
print("PACKAGING REQUIREMENT CONTROL")
print("-" * 70)

print(f"Required Project Files : {len(required_files)}")
print(f"Available Project Files: {len(available_files)}")
print(f"Missing Project Files  : {len(missing_files)}")

if missing_files:
    print("Missing Files:")
    for item in missing_files:
        print(f"  - {item}")

# ----------------------------------------------------------------------
# FINAL PACKAGING STATUS
# ----------------------------------------------------------------------

print("-" * 70)

if len(missing_files) == 0:
    packaging_status = "PASS"
else:
    packaging_status = "CHECK REQUIRED"

print(f"FINAL PACKAGING STATUS : {packaging_status}")

print("=" * 70)

if packaging_status == "PASS":
    print("PHASE 11 — PART 4 — CELL 1: COMPLETED ✓")
    print("FINAL PROJECT PACKAGING CONTROL: PASS ✓")
else:
    print("PHASE 11 — PART 4 — CELL 1: CHECK REQUIRED")

print("=" * 70)
```

### Phase 11 Part 4 Cell 2:
### Final Project Packing Creation

```python
# ======================================================================
# PHASE 11 — PART 4 — CELL 2
# FINAL PROJECT PACKAGE CREATION
# ======================================================================

import os
import zipfile

print("=" * 70)
print("PHASE 11 — PART 4 — CELL 2")
print("FINAL PROJECT PACKAGE CREATION")
print("=" * 70)

# ----------------------------------------------------------------------
# FILE PATHS
# ----------------------------------------------------------------------

excel_file = "/content/AR_AP_Analytics_Dashboard_FINAL.xlsx"

notebook_file = "/content/AR_AP_Analytics_Dashboard_Step_10_05_Part_11.ipynb"

package_file = "/content/AR_AP_Analytics_Dashboard_FINAL_PACKAGE.zip"

# ----------------------------------------------------------------------
# REQUIRED FILE CONTROL
# ----------------------------------------------------------------------

required_files = {
    "Final Excel": excel_file,
    "Final Notebook": notebook_file
}

print("-" * 70)
print("PACKAGE SOURCE FILE CONTROL")
print("-" * 70)

all_available = True

for label, file_path in required_files.items():
    if os.path.exists(file_path):
        size = os.path.getsize(file_path)
        print(f"{label:<20}: AVAILABLE ✓ | {size:,} bytes")
    else:
        print(f"{label:<20}: MISSING ✗")
        all_available = False

# ----------------------------------------------------------------------
# STOP IF ANY REQUIRED FILE IS MISSING
# ----------------------------------------------------------------------

if not all_available:
    raise FileNotFoundError(
        "One or more required project files are missing."
    )

# ----------------------------------------------------------------------
# CREATE FINAL ZIP PACKAGE
# ----------------------------------------------------------------------

print("-" * 70)
print("CREATING FINAL PROJECT PACKAGE")
print("-" * 70)

with zipfile.ZipFile(
    package_file,
    mode="w",
    compression=zipfile.ZIP_DEFLATED
) as zipf:

    zipf.write(
        excel_file,
        arcname="AR_AP_Analytics_Dashboard_FINAL.xlsx"
    )

    zipf.write(
        notebook_file,
        arcname="AR_AP_Analytics_Dashboard_Step_10_05_Part_11.ipynb"
    )

# ----------------------------------------------------------------------
# PACKAGE VALIDATION
# ----------------------------------------------------------------------

print("-" * 70)
print("FINAL PACKAGE VALIDATION")
print("-" * 70)

if not os.path.exists(package_file):
    raise FileNotFoundError(
        "Final package was not created."
    )

package_size = os.path.getsize(package_file)

with zipfile.ZipFile(package_file, "r") as zipf:
    package_contents = zipf.namelist()

expected_contents = [
    "AR_AP_Analytics_Dashboard_FINAL.xlsx",
    "AR_AP_Analytics_Dashboard_Step_10_05_Part_11.ipynb"
]

print(f"Package Exists       : PASS ✓")
print(f"Package Size         : {package_size:,} bytes")
print(f"Files Inside Package : {len(package_contents)}")

print("-" * 70)
print("PACKAGE CONTENT CONTROL")
print("-" * 70)

for filename in expected_contents:
    if filename in package_contents:
        print(f"{filename:<55}: PASS ✓")
    else:
        print(f"{filename:<55}: FAIL ✗")

# ----------------------------------------------------------------------
# FINAL STATUS
# ----------------------------------------------------------------------

package_control_pass = (
    len(package_contents) == 2
    and all(filename in package_contents for filename in expected_contents)
)

print("-" * 70)

if package_control_pass:
    print("PACKAGE CONTENT STATUS : PASS ✓")
    print("FINAL PROJECT PACKAGE  : READY ✓")
else:
    print("PACKAGE CONTENT STATUS : FAIL ✗")
    print("FINAL PROJECT PACKAGE  : CHECK REQUIRED")

print("=" * 70)
print("PHASE 11 — PART 4 — CELL 2: COMPLETED ✓")
print("=" * 70)
```
